```python
import pandas as pd

# Load your cleaned data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/Airbnb_Open_Data_Cleaned.csv')
print(df.shape)
df.head()
```

```python
# Apply the same filter as before to focus on listings with high review rates and low occupancy
df['occupancy_proxy'] = (365 - df['availability 365']) / 365
df['estimated_revenue'] = df['occupancy_proxy'] * df['price'] * 365
df['potential_gain_at_70pct'] = 0.70 * df['price'] * 365 - df['estimated_revenue']
```

```python
# Filter to unused revenue segment
segment = df[(df['review rate number'] >= 4.5) & (df['occupancy_proxy'] < 0.5)].copy()
```

```python
# Keep only useful columns
segment = segment[[
    'id', 'NAME', 'neighbourhood group', 'neighbourhood', 
    'lat', 'long', 'price', 'review rate number', 'number of reviews',
    'occupancy_proxy', 'estimated_revenue', 'potential_gain_at_70pct'
]]
```

```python
# Sort by potential gain
segment = segment.sort_values('potential_gain_at_70pct', ascending=False)
```

```python
# Save new file
segment.to_csv('data/unused_revenue_segment.csv', index=False)
print(f"Exported {len(segment)} listings to prototype/data/unused_revenue_segment.csv")
```
